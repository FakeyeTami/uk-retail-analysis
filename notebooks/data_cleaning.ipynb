```python
import pandas as pd
from IPython.display import display

pd.options.display.max_rows = 100
pd.options.display.max_columns = 100
```

```python
# Load the three raw CSV files
print("\n--- Loading datasets ---\n")

volume = pd.read_csv("../data/raw/uk_retail_volume_monthly.csv")
value = pd.read_csv("../data/raw/uk_retail_value_monthly.csv")
commodity = pd.read_csv("../data/raw/uk_retail_commodity.csv")
```

```python
# Inspecting Raw Datasets
print("\n--- Dataset Overview ---\n")

for name, df in [
    ("volume", volume),
    ("value", value),
    ("commodity", commodity),
]:
    print(f"\n--- {name} ---")
    print(f"Shape: {df.shape}")
    print(f"Columns: {df.columns.tolist()}")
    df.info()


display(volume.iloc[:20])
display(value.iloc[:20])
display(commodity.iloc[:20])
```

```python
# Extract Metadata
def extract_metadata(df, start_row, end_row):
    return df.iloc[start_row:end_row].copy()


volume_metadata = extract_metadata(volume, 0, 11)
value_metadata = extract_metadata(value, 0, 11)
commodity_metadata = extract_metadata(commodity, 0, 12)

display(volume_metadata)
display(value_metadata)
display(commodity_metadata)
```

```python
# Inspecting Metadata
def inspect_metadata(metadata):
    for i in range(len(metadata)):
        print(i, metadata.iloc[i].tolist())


print("\n--- Inspect Volume Metadata ---")
inspect_metadata(volume_metadata)

print("\n--- Inspect Value Metadata ---")
inspect_metadata(value_metadata)

print("\n--- Inspect Commodity Metadata ---")
inspect_metadata(commodity_metadata)
```

```python
# Transform Retail Metadate
def transform_retail_metadata(metadata):
    categories = pd.DataFrame({
        "category_name": metadata.iloc[6, 1:].values,
        "sales_2023": metadata.iloc[7, 1:].values,
        "agg_sic": metadata.iloc[8, 1:].values,
        "percentage_weight": metadata.iloc[9, 1:].values,
        "dataset_code": metadata.iloc[10, 1:].values,
    })

    return categories


volume_categories = transform_retail_metadata(volume_metadata)
value_categories = transform_retail_metadata(value_metadata)

display(volume_categories)
display(value_categories)
```

```python
# Transform Commodity Metadata
def transform_commodity_metadata(metadata):
    commodities = pd.DataFrame({
        "commodity_name": metadata.iloc[9, 1:].values,
        "avg_sales_weekly": metadata.iloc[10, 1:].values,
        "agg_sic": metadata.iloc[11, 1:].values,
    })

    return commodities


commodities = transform_commodity_metadata(commodity_metadata)

display(commodities)
```

```python
# Compare Raw Metadata
print(
    "Volume and value category names match:",
    volume_categories["category_name"].equals(
        value_categories["category_name"]
    )
)

print("Volume categories:", len(volume_categories))
print("Value categories:", len(value_categories))
print("Commodities:", len(commodities))
```

```python
# Cleaning Retail and Commodity Metadata
def clean_retail_metadata(df):
    df = df.copy()

    df["category_name"] = (
        df["category_name"]
        .astype("string")
        .str.strip()
    )

    df["sales_2023"] = (
        df["sales_2023"]
        .astype("string")
        .str.replace(r"[£(),]", "", regex=True)
        .str.replace("m", "", case=False)
        .str.strip()
    )

    df["sales_2023"] = pd.to_numeric(
        df["sales_2023"],
        errors="coerce"
    )

    df["percentage_weight"] = pd.to_numeric(
        df["percentage_weight"],
        errors="coerce"
    )

    df["agg_sic"] = (
        df["agg_sic"]
        .astype("string")
        .str.strip()
    )

    df["dataset_code"] = (
        df["dataset_code"]
        .astype("string")
        .str.strip()
    )

    return df


volume_categories_clean = clean_retail_metadata(
    volume_categories
)

value_categories_clean = clean_retail_metadata(
    value_categories
)

display(volume_categories_clean)
display(value_categories_clean)

def clean_commodity_metadata(df):
    df = df.copy()

    df["commodity_name"] = (
        df["commodity_name"]
        .astype("string")
        .str.replace(
            r"\s*\[Note \d+\]",
            "",
            regex=True
        )
        .str.strip()
    )

    df["avg_sales_weekly"] = (
        df["avg_sales_weekly"]
        .astype("string")
        .str.replace(",", "", regex=True)
        .str.strip()
    )

    df["avg_sales_weekly"] = pd.to_numeric(
        df["avg_sales_weekly"],
        errors="coerce"
    )

    df["agg_sic"] = (
        df["agg_sic"]
        .astype("string")
        .str.strip()
    )

    return df


commodity_data_clean = clean_commodity_metadata(
    commodities
)

display(commodity_data_clean)
```

```python
# Comparing Retail Categories
print(
    "Volume categories:",
    len(volume_categories_clean)
)

print(
    "Value categories:",
    len(value_categories_clean)
)

print(
    "Same category names:",
    volume_categories_clean["category_name"].equals(
        value_categories_clean["category_name"]
    )
)

print(
    "Same AGG/SIC:",
    volume_categories_clean["agg_sic"].equals(
        value_categories_clean["agg_sic"]
    )
)

print(
    "Same percentage weights:",
    volume_categories_clean["percentage_weight"].equals(
        value_categories_clean["percentage_weight"]
    )
)

print(
    "Same 2023 sales:",
    volume_categories_clean["sales_2023"].equals(
        value_categories_clean["sales_2023"]
    )
)

print(
    "Same dataset codes:",
    volume_categories_clean["dataset_code"].equals(
        value_categories_clean["dataset_code"]
    )
)

comparison = pd.DataFrame({
    "category": volume_categories_clean["category_name"],
    "volume_dataset_code": volume_categories_clean["dataset_code"],
    "value_dataset_code": value_categories_clean["dataset_code"],
})

display(comparison)
```

```python
# Creating Retail Category Table
retail_category = volume_categories_clean[
    [
        "category_name",
        "agg_sic",
        "percentage_weight",
        "sales_2023",
    ]
].copy()

retail_category.insert(0, "category_id", range(1, len(retail_category) + 1))

display(retail_category)

# Validate Table
print("Rows:", len(retail_category))
print("Duplicate category names:", retail_category["category_name"].duplicated().sum())
print("\nMissing values:", retail_category.isnull().sum())
print("Unique category IDs:", retail_category["category_id"].nunique())
print("Category IDs unique:", retail_category["category_id"].is_unique)

# Export Retail Category Table
retail_category.to_csv("../data/processed/retail_category.csv", index=False)
print("Retail category table exported.")
```

```python
# Extract Raw Retail Data
def extract_monthly_data(df, start_row=11, end_row=472):
    return df.iloc[start_row:end_row + 1].copy()

volume_monthly_raw = extract_monthly_data(volume)
value_monthly_raw = extract_monthly_data(value)

display(volume_monthly_raw.head())
display(value_monthly_raw.head())
```

```python
# Inspect Raw Monthly Data
print("Volume: ", volume_monthly_raw.shape)
print("Value: ", value_monthly_raw.shape)

# Inspect first and last periods
print(volume_monthly_raw.iloc[0, 0])
print(volume_monthly_raw.iloc[-1, 0])

print(value_monthly_raw.iloc[0, 0])
print(value_monthly_raw.iloc[-1, 0])
```

```python
# Transform time_period to dim_date
def transform_time_period(df):
    df = df.copy()

    df = df.rename(
        columns={
            df.columns[0]: "time_period"
        }
    )

    df["time_period"] = pd.to_datetime(
        df["time_period"],
        format="%Y %b",
        errors="coerce"
    )

    return df


volume_monthly_raw = transform_time_period(
    volume_monthly_raw
)

value_monthly_raw = transform_time_period(
    value_monthly_raw
)

# Create date_id
def create_date_id(df):
    df = df.copy()

    df["date_id"] = (
        df["time_period"].dt.year * 100
        + df["time_period"].dt.month
    )

    return df


volume_monthly_raw = create_date_id(
    volume_monthly_raw
)

value_monthly_raw = create_date_id(
    value_monthly_raw
)

# Validate changes
display(
    volume_monthly_raw[
        ["date_id", "time_period"]
    ].head()
)

display(
    value_monthly_raw[
        ["date_id", "time_period"]
    ].head()
)

# Check Uniqueness
print(volume_monthly_raw["date_id"].is_unique)
print(value_monthly_raw["date_id"].is_unique)
```

```python
# Rename Monthly Category Columns
def rename_category_columns(df, metadata):
    df = df.copy()

    category_names = metadata.iloc[6, 1:].tolist()

    monthly_category_columns = df.columns[1:-1]

    if len(monthly_category_columns) != len(category_names):
        raise ValueError(
            f"Category mismatch: "
            f"{len(monthly_category_columns)} monthly columns vs "
            f"{len(category_names)} metadata categories"
        )

    column_mapping = dict(
        zip(monthly_category_columns, category_names)
    )

    df = df.rename(columns=column_mapping)

    return df

volume_monthly_raw = rename_category_columns(
    volume_monthly_raw,
    volume_metadata
)

value_monthly_raw = rename_category_columns(
    value_monthly_raw,
    value_metadata
)

print("Volume columns:", len(volume_monthly_raw.columns))
print("Value columns:", len(value_monthly_raw.columns))

display(volume_monthly_raw.head())
display(value_monthly_raw.head())
```

```python
# Transform Monthly Data
def transform_monthly_data(df, value_name):
    df = df.copy()

    id_columns = [
        "date_id",
        "time_period"
    ]

    value_columns = [
        col
        for col in df.columns
        if col not in id_columns
    ]

    df = df.melt(
        id_vars=id_columns,
        value_vars=value_columns,
        var_name="category_name",
        value_name=value_name
    )

    return df


volume_monthly = transform_monthly_data(volume_monthly_raw, "volume_index")
value_monthly = transform_monthly_data(value_monthly_raw, "sales_value")

display(volume_monthly.head(20))
display(value_monthly.head(20))
```

```python
# Match Category Names and ID
category_lookup = retail_category[
    [
        "category_id",
        "category_name"
    ]
].copy()


volume_monthly = volume_monthly.merge(
    category_lookup,
    on="category_name",
    how="left",
    validate="many_to_one"
)

value_monthly = value_monthly.merge(
    category_lookup,
    on="category_name",
    how="left",
    validate="many_to_one"
)

# Validate
print("Missing volume category IDs:", volume_monthly["category_id"].isna().sum())
print("Missing value category IDs:", value_monthly["category_id"].isna().sum())
```

```python
# Add Dataset Codes

volume_monthly = volume_monthly.merge(
    volume_categories_clean[
        [
            "category_name",
            "dataset_code"
        ]
    ],
    on="category_name",
    how="left",
    validate="many_to_one"
)

volume_monthly = volume_monthly.rename(
    columns={
        "dataset_code": "volume_code"
    }
)

value_monthly = value_monthly.merge(
    value_categories_clean[
        [
            "category_name",
            "dataset_code"
        ]
    ],
    on="category_name",
    how="left",
    validate="many_to_one"
)

value_monthly = value_monthly.rename(
    columns={
        "dataset_code": "value_code"
    }
)
```

```python
# Create Volume and Value Table

volume_monthly = volume_monthly[
    [
        "date_id",
        "time_period",
        "category_id",
        "volume_code",
        "volume_index",
    ]
]

value_monthly = value_monthly[
    [
        "date_id",
        "time_period",
        "category_id",
        "value_code",
        "sales_value",
    ]
]

volume_monthly["volume_index"] = pd.to_numeric(
    volume_monthly["volume_index"],
    errors="coerce"
)

value_monthly["sales_value"] = pd.to_numeric(
    value_monthly["sales_value"],
    errors="coerce"
)

display(volume_monthly.head())
display(value_monthly.head())

print(volume_monthly.shape)
print(value_monthly.shape)

print(
    "Duplicate volume keys:",
    volume_monthly.duplicated(
        ["date_id", "category_id"]
    ).sum()
)

print(
    "Duplicate value keys:",
    value_monthly.duplicated(
        ["date_id", "category_id"]
    ).sum()
)
```

```python
# Validate

print("--- Volume validation ---")
print("Rows:", len(volume_monthly))
print("Missing values:")
print(volume_monthly.isna().sum())
print(
    "Duplicate keys:",
    volume_monthly.duplicated(["date_id", "category_id"]).sum()
)

print("\n--- Value validation ---")
print("Rows:", len(value_monthly))
print("Missing values:")
print(value_monthly.isna().sum())
print(
    "Duplicate keys:",
    value_monthly.duplicated(["date_id", "category_id"]).sum()
)

print(
    "Volume missing category IDs:",
    volume_monthly["category_id"].isna().sum()
)

print(
    "Value missing category IDs:",
    value_monthly["category_id"].isna().sum()
)

print(volume_monthly.columns.tolist())
print(value_monthly.columns.tolist())
```
